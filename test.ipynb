```python
def checkB(s):
    if len(s) == 1:
        return True

    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:
            nxt = (
                s[:i]
                + [1 - s[i]]
                + s[i + 2:]
            )

            if checkB(nxt):
                return True

    return False
```

```python
t = int(input())

for _ in range(t):
    n = int(input())
    s = input()
    output = 0
    arr = [int(ch) for ch in s]

    for i in range(n):
        for j in range(i + 1, len(arr) + 1):
            new = arr[i:j]
            if checkB(new):
                output+=1

    print(output)
```
