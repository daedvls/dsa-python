### Hash Sets

```python
# hashsets
s = set()
print(s)
```

```python
# add item into set - O(1)
s.add(1)
s.add(2)
s.add(3)

print(s)
```

```python
# Lookup if item is in set or not - O(1)
if 1 in s:
    print("yes")
```

```python
# Remove element from set - O(1)
# s.remove(4) # Key error
s.remove(3)
s
```

```python
string = "aaaaaaabbbbbcceeee"
sett = set(string)  # Set construction - O(S) (S is len)

sett
```

### Hash Maps (aka Dictionaries)

```python
d = {'joel':69, 'yo': 2, 'j': 5}

d
```

```python
# add key:val - O(1)
d['br'] = 4
d
```

```python
# check if key is present in hashmap - O(1)

if 'joel' in d:
    print(True)
```

```python
# check the value corresponding to a key - O(1)

print(d['br'])
# Note this assumes that 'br' is a key in the hashmap
```

```python
# loop over key, val pairs - O(n)
for key, val in d.items():
    print(f"{key}:{val}")
```
